```python
x = pd.get_dummies(x, drop_first=True)
```

```python
x.dtypes
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)
```

```python
from sklearn.preprocessing import StandardScaler

SS = StandardScaler()

x_train = SS.fit_transform(x_train)
x_test = SS.transform(x_test)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
LR= LinearRegression()
```

```python
LR.fit(x_train,y_train)
```

```python
y_pred = LR.predict(x_test)
```

```python
from sklearn.metrics import mean_squared_error,root_mean_squared_error,mean_absolute_error,r2_score
```

```python
mean_squared_error(y_test, y_pred)
```

```python
root_mean_squared_error(y_test, y_pred)
```

```python
mean_absolute_error(y_test, y_pred)
```

```python
r2_score(y_test, y_pred)
```

```python
LR.coef_
```

```python
LR.intercept_
```

```python
from sklearn.linear_model import Lasso
```

```python
L1=Lasso(alpha=1)
```

```python
L1.fit(x_train, y_train)
```

```python
y_pred_L1 = L1.predict(x_test)
```

```python
L1.score(x_test,y_test)
```

```python
mean_squared_error(y_test, y_pred_L1)
```

```python
root_mean_squared_error(y_test, y_pred_L1)
```

```python
mean_absolute_error(y_test, y_pred_L1)
```

```python
r2_score(y_test, y_pred_L1)
```

```python
from sklearn.linear_model import Ridge
```

```python
L2=Ridge(alpha=0.1)
```

```python
L2.fit(x_train,y_train)
```

```python
L2.score(x_test,y_test)
```

```python
y_pred_L2 = L2.predict(x_test)
```

```python
mean_squared_error(y_test, y_pred_L2)
```

```python
root_mean_squared_error(y_test, y_pred_L2)
```

```python
mean_absolute_error(y_test, y_pred_L2)
```

```python
r2_score(y_test, y_pred_L2)
```
